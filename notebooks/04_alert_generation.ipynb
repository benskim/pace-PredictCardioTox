# 04 — Alert Generation

Generate validation alerts from absolute QTc thresholds and changes from baseline.

```python
import pandas as pd

from ecg_analytics.alerts import AlertRule, generate_qtc_alerts
```

```python
follow_up = pd.DataFrame({
    'subject_id': ['A', 'B', 'C', 'D'],
    'visit': ['day-7', 'day-7', 'day-7', 'day-7'],
    'baseline_qtc_ms': [405.0, 440.0, 410.0, 470.0],
    'qtc_ms': [430.0, 505.0, 475.0, 498.0],
})
follow_up
```

```python
rule = AlertRule(absolute_qtc_ms=500.0, delta_qtc_ms=60.0)
alerts = generate_qtc_alerts(follow_up, rule)
alerts
```

```python
if alerts.empty:
    print('No QTc alerts generated.')
else:
    display(alerts[['subject_id', 'visit', 'qtc_ms', 'delta_qtc_ms', 'absolute_qtc_alert', 'delta_qtc_alert']])
```
