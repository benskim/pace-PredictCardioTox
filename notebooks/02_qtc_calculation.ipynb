# 02 — QTc Calculation

Validate QT correction formulas on representative QT and RR intervals before applying them to annotated beats.

```python
import pandas as pd

from ecg_analytics.qtc import qtc_bazett, qtc_fridericia, qtc_framingham, qtc_hodges
```

```python
measurements = pd.DataFrame({
    'subject_id': ['example-1', 'example-2', 'example-3'],
    'qt_ms': [390.0, 420.0, 460.0],
    'rr_ms': [800.0, 1000.0, 1200.0],
})
measurements['heart_rate_bpm'] = 60000.0 / measurements['rr_ms']
measurements
```

```python
measurements['qtc_bazett_ms'] = qtc_bazett(measurements['qt_ms'], measurements['rr_ms'])
measurements['qtc_fridericia_ms'] = qtc_fridericia(measurements['qt_ms'], measurements['rr_ms'])
measurements['qtc_framingham_ms'] = qtc_framingham(measurements['qt_ms'], measurements['rr_ms'])
measurements['qtc_hodges_ms'] = qtc_hodges(measurements['qt_ms'], measurements['heart_rate_bpm'])
measurements
```

```python
ax = measurements.set_index('subject_id')[[
    'qtc_bazett_ms', 'qtc_fridericia_ms', 'qtc_framingham_ms', 'qtc_hodges_ms'
]].plot(kind='bar', figsize=(10, 4), title='QTc formula comparison')
ax.set_ylabel('QTc (ms)')
```
