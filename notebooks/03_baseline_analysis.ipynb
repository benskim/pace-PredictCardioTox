# 03 — Baseline Analysis

Summarize baseline QTc distributions by subject or cohort to support analytics validation.

```python
import pandas as pd

from ecg_analytics.baseline import baseline_summary
```

```python
baseline_measurements = pd.DataFrame({
    'subject_id': ['A', 'A', 'A', 'B', 'B', 'C'],
    'visit': ['baseline'] * 6,
    'qtc_ms': [401.0, 408.0, 412.0, 430.0, 435.0, 455.0],
})
baseline_measurements
```

```python
summary = baseline_summary(baseline_measurements, group_columns='subject_id', qtc_column='qtc_ms')
summary
```

```python
ax = summary.plot(x='subject_id', y='mean', kind='bar', yerr='std', legend=False, title='Mean baseline QTc by subject')
ax.set_ylabel('QTc (ms)')
```
