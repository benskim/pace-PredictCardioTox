Deep-dive into measurement errors: Bland-Altman plots, error

```python
import sys
sys.path.insert(0, '../src')
import numpy as np
```

```python
from ecg_analytics.validation import qt_metrics, qtc_metrics, delineation_metrics
from ecg_analytics.visualization import plot_bland_altman, plot_error_distribution
```

## Bland-Altman Analysis

```python
np.random.seed(42)
predicted = np.random.normal(400, 20, 100)
reference = predicted + np.random.normal(2, 8, 100)

fig = plot_bland_altman(predicted, reference, title='QT Bland-Altman')
```

## Error Distribution

```python
errors = predicted - reference
fig = plot_error_distribution(errors, title='QT Error Distribution')
```

## Summary Metrics

```python
metrics = qt_metrics(predicted, reference)
import pandas as pd
pd.DataFrame([metrics], index=['QT Measurement'])
```

## Outlier Identification

```python
threshold = 2 * np.std(errors)
outliers = np.where(np.abs(errors) > threshold)[0]
print(f'Outliers (>{threshold:.1f} ms): {len(outliers)} of {len(errors)} ({100*len(outliers)/len(errors):.1f}%)')
```
