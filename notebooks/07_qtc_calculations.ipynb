# 07 — QTc Calculations

Compute all four QTc corrections simultaneously.
Default reporting formula: **Fridericia**.

```python
import sys
sys.path.insert(0, '../src')
import numpy as np
import pandas as pd
```

```python
from ecg_analytics.qtc import compute_all_qtc, qtc_fridericia
```

## Single-Value Computation

```python
qt_ms = 400.0
rr_ms = 800.0  # 75 bpm
all_qtc = compute_all_qtc(qt_ms, rr_ms)
for name, val in all_qtc.items():
    print(f'{name:>12s}: {float(val):.2f} ms')
```

## Vectorised (Per-Beat) Computation

```python
qt_array = np.array([380, 390, 400, 410, 420], dtype=float)
rr_array = np.array([900, 850, 800, 750, 700], dtype=float)

df = pd.DataFrame({'qt_ms': qt_array, 'rr_ms': rr_array})
all_qtc = compute_all_qtc(df['qt_ms'], df['rr_ms'])
for name, vals in all_qtc.items():
    df[f'qtc_{name}'] = vals
df
```

## Fridericia as Default

```python
print('Default (Fridericia):', qtc_fridericia(400, 800))
```
