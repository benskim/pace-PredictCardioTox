# 15 — Beat-to-Beat QT Variability

This notebook examines beat-level QT consistency and computes
the **Beat Stability Score**.

Metrics: Mean QT, Median QT, SD, IQR, Coefficient of Variation.


```python
import numpy as np
import matplotlib.pyplot as plt

from ecg_analytics.confidence.beat_stability import (
    beat_qt_stats,
    beat_stability_score,
)
```

## Simulated beat-level QT intervals


```python
np.random.seed(123)

# Stable subject
stable_qts = np.random.normal(400, 2, 20)

# Unstable subject
unstable_qts = np.random.normal(400, 15, 20)

print('Stable QTs:', np.round(stable_qts[:5], 1), '...')
print('Unstable QTs:', np.round(unstable_qts[:5], 1), '...')
```

## Beat statistics


```python
stats_stable = beat_qt_stats(stable_qts)
stats_unstable = beat_qt_stats(unstable_qts)

print('Stable subject:')
for k, v in stats_stable.items():
    print(f'  {k}: {v:.2f}')

print('\nUnstable subject:')
for k, v in stats_unstable.items():
    print(f'  {k}: {v:.2f}')
```

## Beat Stability Score


```python
result_stable = beat_stability_score(stable_qts)
result_unstable = beat_stability_score(unstable_qts)

print(f'Stable: score={result_stable.stability_score:.1f}, '
      f'CV={result_stable.cv:.4f}, IQR={result_stable.iqr_qt_ms:.1f}')
print(f'Unstable: score={result_unstable.stability_score:.1f}, '
      f'CV={result_unstable.cv:.4f}, IQR={result_unstable.iqr_qt_ms:.1f}')
```

## Visualization


```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

for ax, qts, label, score in [
    (axes[0], stable_qts, 'Stable', result_stable.stability_score),
    (axes[1], unstable_qts, 'Unstable', result_unstable.stability_score),
]:
    ax.plot(qts, 'o-', markersize=4)
    ax.axhline(np.mean(qts), color='red', linestyle='--', label=f'Mean: {np.mean(qts):.1f}')
    ax.fill_between(
        range(len(qts)),
        np.mean(qts) - np.std(qts),
        np.mean(qts) + np.std(qts),
        alpha=0.15, color='red', label='±1 SD',
    )
    ax.set_title(f'{label} (score: {score:.0f})')
    ax.set_xlabel('Beat')
    ax.set_ylabel('QT (ms)')
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```
