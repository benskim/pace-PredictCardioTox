## Synthetic Cross-Dataset Comparison

```python
datasets = ['QTDB', 'LUDB', 'CSE']
np.random.seed(42)
rows = []
for ds in datasets:
    pred = np.random.normal(400, 20, 50)
    ref = pred + np.random.normal(0, 5, 50)
    m = qt_metrics(pred, ref)
    m['dataset'] = ds
    rows.append(m)

import pandas as pd
pd.DataFrame(rows).set_index('dataset')
```
