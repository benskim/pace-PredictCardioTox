Load and inspect records from the PhysioNet QT Database, LUDB, NSTDB, INCART, and PTB-XL datasets via the unified `ECGRecord` interface.

```python
from ecg_analytics.datasets import QTDBDataset, LUDBDataset, NSTDBDataset, INCARTDataset, PTBXLDataset
```

```python
from ecg_analytics.visualization import plot_multi_lead
```

## PhysioNet QT Database

```python
qtdb = QTDBDataset(data_dir='../data/physionet')
# qtdb.download(records=['sel100', 'sel102'])  # uncomment to download
print('QTDB records:', qtdb.list_records()[:10])
```

```python
rec = qtdb.load_record('sel100')
print(f'Record: {rec.record_id}, fs={rec.fs} Hz, duration={rec.duration_s:.1f}s, leads={rec.lead_names}')
```

## LUDB

```python
ludb = LUDBDataset(data_dir='../data/physionet')
ludb.download(records=['1', '2'])  # uncomment to download
print('LUDB records:', ludb.list_records()[:10])
rec = ludb.load_record('1')
print(f'Record: {rec.record_id}, fs={rec.fs} Hz, duration={rec.duration_s:.1f}s, leads={rec.lead_names}')
```

## MIT-BIH Noise Stress Test Database (NSTDB)

```python
nstdb = NSTDBDataset(data_dir='../data/physionet')
nstdb.download(records=['118e00', '118e06'])  # uncomment to download
print('NSTDB records:', nstdb.list_records()[:10])
```

```python
nstdb_record_ids = nstdb.list_records()
if nstdb_record_ids:
    rec = nstdb.load_record(nstdb_record_ids[0])
    print(f'Record: {rec.record_id}, fs={rec.fs} Hz, duration={rec.duration_s:.1f}s, leads={rec.lead_names}')
```

## INCART Database

```python
incart = INCARTDataset(data_dir='../data/physionet')
incart.download(records=['I01', 'I02'])  # uncomment to download
print('INCART records:', incart.list_records()[:10])
```

```python
incart_record_ids = incart.list_records()
if incart_record_ids:
    rec = incart.load_record(incart_record_ids[0])
    print(f'Record: {rec.record_id}, fs={rec.fs} Hz, duration={rec.duration_s:.1f}s, leads={rec.lead_names}')
```
