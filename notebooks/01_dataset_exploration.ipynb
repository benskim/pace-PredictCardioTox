## Signal Inspection

```python
from ecg_analytics.visualization import plot_ecg_signal
import numpy as np

# Example with synthetic data
fs = 250.0
t = np.arange(0, 5, 1/fs)
signal = 0.8 * np.sin(2 * np.pi * 1.2 * t) + 0.1 * np.random.randn(len(t))
fig = plot_ecg_signal(signal, fs, title='Synthetic ECG')
```
