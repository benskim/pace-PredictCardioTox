# 05 — Geometric Tangent Method for T-end

The tangent method determines T-end by fitting a line at the point of
maximum downslope on the trailing T-wave edge, then finding its
intersection with the isoelectric baseline.

```python
import sys
sys.path.insert(0, '../src')
import numpy as np
import matplotlib.pyplot as plt
```

```python
from ecg_analytics.qt import tangent_t_end, find_t_peak, extract_t_wave_region
```

## Construct a Synthetic Beat

```python
fs = 500.0
t = np.arange(0, 1.0, 1/fs)
qrs = 1.5 * np.exp(-((t - 0.2)**2) / (2 * 0.004**2))
t_wave = 0.4 * np.exp(-((t - 0.45)**2) / (2 * 0.02**2))
signal = qrs + t_wave

r_peak = int(0.2 * fs)
t_start, t_end_search = extract_t_wave_region(signal, r_peak, fs)
t_peak_idx = find_t_peak(signal, t_start, t_end_search)
print(f'T-peak at sample {t_peak_idx}')
```

## Apply Tangent Method

```python
t_end_idx = tangent_t_end(signal, t_peak_idx, fs, baseline=0.0)
print(f'T-end at sample {t_end_idx} ({t_end_idx/fs*1000:.1f} ms)')
```

```python
fig, ax = plt.subplots(figsize=(12, 4))
time = np.arange(len(signal)) / fs * 1000
ax.plot(time, signal, 'b-', lw=1)
ax.axvline(t_peak_idx/fs*1000, color='green', ls='--', label='T-peak')
if t_end_idx:
    ax.axvline(t_end_idx/fs*1000, color='red', ls='--', label='T-end (tangent)')
ax.axhline(0, color='gray', ls=':', alpha=0.5)
ax.legend()
ax.set_xlabel('Time (ms)')
ax.set_title('Tangent Method T-end Determination')
plt.tight_layout()
```
