# 03 — Wave Delineation (1-D U-Net)

Demonstrate the wave delineation pipeline using the 1-D U-Net
segmentation model. The model classifies each sample as
background, P-wave, QRS, or T-wave.

```python
import sys
sys.path.insert(0, '../src')
import numpy as np
import torch
```

```python
from ecg_analytics.delineation import UNet1D, delineate, WAVE_CLASSES
from ecg_analytics.delineation.labels import encode_annotations, decode_mask
```

## Model Architecture

```python
model = UNet1D(in_channels=1, n_classes=4, base_filters=32, depth=4)
print(model)
print(f'\nParameters: {sum(p.numel() for p in model.parameters()):,}')
```

## Inference on Synthetic Signal

```python
fs = 250.0
t = np.arange(0, 5, 1/fs)
signal = np.sin(2 * np.pi * 1.2 * t).astype(np.float32)

result = delineate(signal, model=model)
print(f'Mask shape: {result.mask.shape}')
print(f'Unique classes: {np.unique(result.mask)}')
print(f'Regions: {result.regions}')
```
