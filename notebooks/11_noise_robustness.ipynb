# 11 — Noise Robustness Evaluation

Evaluate T-end and QT accuracy degradation under controlled noise
conditions at SNR levels: 24, 18, 12, 6, 0 dB.

```python
import sys
sys.path.insert(0, '../src')
import numpy as np
```

```python
from ecg_analytics.preprocessing.noise import NOISE_FUNCTIONS, STANDARD_SNR_LEVELS, inject_noise
from ecg_analytics.visualization import plot_noise_comparison, plot_noise_degradation, plot_error_heatmap
```

## Noise Types

```python
print('Noise types:', list(NOISE_FUNCTIONS.keys()))
print('SNR levels:', STANDARD_SNR_LEVELS)
```

## Visual Comparison

```python
fs = 250.0
t = np.arange(0, 5, 1/fs)
clean = np.sin(2 * np.pi * 1.2 * t)

for noise_type in NOISE_FUNCTIONS:
    noisy = inject_noise(clean, fs, noise_type, snr_db=12.0, rng=np.random.default_rng(42))
    fig = plot_noise_comparison(clean, noisy, fs, noise_label=noise_type, snr_db=12.0)
```

## Degradation Curves

```python
# Synthetic degradation data
mae_values = {}
for noise_type in NOISE_FUNCTIONS:
    mae_values[noise_type] = [2 + (24 - s) * 0.5 + np.random.rand() for s in STANDARD_SNR_LEVELS]

fig = plot_noise_degradation(STANDARD_SNR_LEVELS, mae_values, metric_name='QT MAE (ms)')
```

## Error Heatmap

```python
noise_types = list(NOISE_FUNCTIONS.keys())
errors = np.array([[mae_values[nt][i] for i in range(len(STANDARD_SNR_LEVELS))] for nt in noise_types])
fig = plot_error_heatmap(noise_types, STANDARD_SNR_LEVELS, errors)
```
