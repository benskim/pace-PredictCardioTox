## Cohort-Level Drift Summary

```python
np.random.seed(42)
n_subjects, n_timepoints = 20, 5
rows = []
for s in range(n_subjects):
    bl = np.random.normal(420, 20)
    for tp in range(n_timepoints):
        qtc = bl + tp * np.random.normal(5, 10)
        rows.append({'subject_id': f'S{s:03d}', 'time_point': tp, 'qtc_fridericia': qtc})
df = pd.DataFrame(rows)

summary = subject_drift_summary(df, baseline_time=0)
summary
```

```python
# Flag subjects with significant drift
summary[summary['exceeds_threshold']]
```
