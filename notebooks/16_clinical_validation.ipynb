# 16 — Clinical Validation Package

This notebook demonstrates the clinical validation utilities:

- **Bland–Altman analysis** with bias and limits of agreement
- **Coverage analysis** at ±5, ±10, ±20 ms
- Publication-quality figures


```python
import numpy as np
import matplotlib.pyplot as plt

from ecg_analytics.validation.clinical import (
    bland_altman_analysis,
    plot_clinical_bland_altman,
    plot_coverage_bars,
)
```

## Simulate paired measurements


```python
np.random.seed(42)
n = 200
reference = np.random.normal(420, 20, n)
predicted = reference + np.random.normal(1.5, 6, n)  # slight bias + noise
```

## Bland–Altman Analysis


```python
ba = bland_altman_analysis(predicted, reference)
print(f'Bias: {ba.bias:.2f} ms')
print(f'SD: {ba.sd:.2f} ms')
print(f'LoA: [{ba.loa_lower:.2f}, {ba.loa_upper:.2f}] ms')
print(f'Coverage ±5 ms: {ba.coverage_5ms:.1%}')
print(f'Coverage ±10 ms: {ba.coverage_10ms:.1%}')
print(f'Coverage ±20 ms: {ba.coverage_20ms:.1%}')
```

## Bland–Altman Plot (publication quality)


```python
fig = plot_clinical_bland_altman(predicted, reference,
                                 title='QT Measurement: Method vs Reference')
plt.show()
```

## Coverage Bar Chart


```python
fig = plot_coverage_bars(ba, title='QT Measurement Coverage')
plt.show()
```

## Multi-condition Comparison

Compare Bland–Altman statistics across noise conditions.


```python
conditions = ['Clean', 'SNR 24 dB', 'SNR 12 dB', 'SNR 6 dB']
noise_levels = [0, 3, 8, 15]  # additional noise SD (ms)

results = []
for cond, noise_sd in zip(conditions, noise_levels):
    noisy_pred = reference + np.random.normal(1.5, 6 + noise_sd, n)
    ba_result = bland_altman_analysis(noisy_pred, reference)
    results.append((cond, ba_result))
    print(f'{cond}: bias={ba_result.bias:.2f}, '
          f'SD={ba_result.sd:.2f}, '
          f'±10ms={ba_result.coverage_10ms:.1%}')
```
