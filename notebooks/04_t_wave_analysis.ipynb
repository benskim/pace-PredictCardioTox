# 04 — T-wave Analysis

Extract T-wave regions, locate T-peaks, and analyze T-wave morphology.

```python
import sys
sys.path.insert(0, '../src')
import numpy as np
```

```python
from ecg_analytics.qt import extract_t_wave_region, find_t_peak
```

## T-wave Region Extraction

```python
# Synthetic ECG with clear T-wave
fs = 250.0
t = np.arange(0, 2, 1/fs)
qrs = np.exp(-((t - 0.5)**2) / (2 * 0.005**2))
t_wave = 0.3 * np.exp(-((t - 0.8)**2) / (2 * 0.03**2))
signal = qrs + t_wave + 0.02 * np.random.randn(len(t))

r_peak = int(0.5 * fs)
t_start, t_end = extract_t_wave_region(signal, r_peak, fs)
print(f'T-wave search region: samples {t_start}-{t_end} ({t_start/fs:.3f}s - {t_end/fs:.3f}s)')
```

```python
t_peak = find_t_peak(signal, t_start, t_end)
if t_peak is not None:
    print(f'T-peak at sample {t_peak} ({t_peak/fs:.3f}s)')
```

## Visualization

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize=(12, 4))
time = np.arange(len(signal)) / fs
ax.plot(time, signal, 'b-', lw=0.8)
ax.axvspan(t_start/fs, t_end/fs, alpha=0.2, color='orange', label='T-wave region')
if t_peak:
    ax.axvline(t_peak/fs, color='red', ls='--', label='T-peak')
ax.legend()
ax.set_xlabel('Time (s)')
ax.set_title('T-wave Region Extraction')
plt.tight_layout()
```
