End-to-end QT measurement: R-peak detection → Q-onset → T-end
(tangent method) → QT interval.

```python
from ecg_analytics.qt import measure_qt_intervals
```

## Synthetic Multi-Beat Signal

```python
fs = 500.0
t = np.arange(0, 5.0, 1/fs)
# Three beats
signal = np.zeros_like(t)
for beat_time in [0.5, 1.5, 2.5, 3.5]:
    qrs = 1.5 * np.exp(-((t - beat_time)**2) / (2 * 0.004**2))
    tw = 0.4 * np.exp(-((t - beat_time - 0.25)**2) / (2 * 0.02**2))
    signal += qrs + tw
signal += 0.02 * np.random.randn(len(t))
```

```python
results = measure_qt_intervals(signal, fs)
for m in results:
    qt = f'{m.qt_ms:.1f} ms' if m.qt_ms else 'N/A'
    rr = f'{m.rr_ms:.1f} ms' if m.rr_ms else 'N/A'
    print(f'R={m.r_peak}, Q={m.q_onset}, T-end={m.t_end}, QT={qt}, RR={rr}')
```
